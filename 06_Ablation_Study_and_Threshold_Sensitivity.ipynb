1. **7-Stage Ablation Configurations**:
   - **C1**: Baseline Dual-LSTM (Single timestep)
   - **C2**: + Sliding Temporal Window ($T = 5$)
   - **C3**: + Transformer Self-Attention Branch (Trans-LSTM)
